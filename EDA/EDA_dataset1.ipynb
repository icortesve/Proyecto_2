```python
print('=== Detección de Inconsistencias en Columnas Categóricas ===')
print()

# Identificar columnas categóricas en el DataFrame
cols_categoricas = df.select_dtypes(include=['category', 'object']).columns

if cols_categoricas.empty:
    print("• No se encontraron columnas de tipo 'category' u 'object' en el DataFrame. ✅")
    print("  Por lo tanto, no hay inconsistencias categóricas que chequear en este paso.")
else:
    print(f"• Se encontraron las siguientes columnas categóricas: {list(cols_categoricas)}")
    print("  Realizando verificación de inconsistencias...")

    # Limpieza de espacios invisibles y correcciones específicas (si fueran necesarias)
    for col in cols_categoricas:
        df[col] = df[col].astype(str).str.strip()

    columnas_con_inconsistencias = []

    for col in cols_categoricas:
        serie_limpia = df[col].str.capitalize() # Normalización temporal para la búsqueda
        conteo = serie_limpia.value_counts()
        valores_validos = conteo[conteo > 1].index.tolist()

        print(f'Columna: {col}')
        inconsistent_values = serie_limpia[~serie_limpia.isin(valores_validos)].unique().tolist()

        if len(inconsistent_values) > 0:
            print(f'• Valores únicos detectados (aparecen solo una vez): {inconsistent_values}')
            print("  Ejemplos de filas con estos valores:")
            display(df[df[col].str.capitalize().isin(inconsistent_values)][[col]])
            columnas_con_inconsistencias.append(col)
        else:
            print(f'• Sin inconsistencias detectadas (todos los valores aparecen más de una vez) ✅')
        print('-' * 40)

    # Devolver las variables de texto remanentes a tipo 'category'
    for col in cols_categoricas:
        df[col] = df[col].astype('category')

    if len(columnas_con_inconsistencias) == 0:
        print("\n✅ Resumen: No se encontraron nuevas inconsistencias o todas las detectadas fueron abordadas.")
    else:
        print(f"\n⚠️ Resumen: Se encontraron inconsistencias en las siguientes columnas: {', '.join(columnas_con_inconsistencias)}.")
        print("         (Estas serán tratadas en los siguientes pasos, como la agrupación en 'Other' o One-Hot Encoding).")
print('-' * 80)
```

  * Verificación de Normalidad

```python
print('=== Verificación de Normalidad de Variables (Test de Shapiro-Wilk) ===\n')

from scipy.stats import shapiro
from pandas.api.types import is_numeric_dtype

# Seleccionar solo columnas numéricas para la prueba de normalidad
variables_para_normalidad = [col for col in df.columns if is_numeric_dtype(df[col]) and col not in ['Id', 'id']]

alpha = 0.05  # Nivel de significancia

print(f"Hipótesis nula (H0): La distribución de la variable es normal.\nNivel de significancia (alpha): {alpha}\n")

for col in variables_para_normalidad:
    stat, p = shapiro(df[col])
    print(f'\n--- Columna: {col} ---')
    print(f'Estadístico de Shapiro-Wilk: {stat:.4f}')
    print(f'Valor p: {p:.4e}') # Formato exponencial para p-values pequeños
    if p > alpha:
        print('Conclusión: No se puede rechazar la hipótesis nula (H0). La columna parece tener una distribución normal. ✅')
    else:
        print('Conclusión: Se rechaza la hipótesis nula (H0). La columna NO sigue una distribución normal. ❌')

print('\n--------------------------------------------------------------------------------')
print('💡 Nota: Para datasets grandes (ej. > 5000 muestras), el test de Shapiro-Wilk puede ser muy sensible a pequeñas desviaciones de la normalidad. En esos casos, la inspección visual (histogramas, Q-Q plots) o otros tests como Kolmogorov-Smirnov pueden ser más útiles.')
```

## Outliers



```python
print('=== Análisis y detección de valores atípicos (Outliers) ===\n')

# Excluir columnas no numéricas y de identificación para el análisis de outliers
variables_analisis = [col for col in df.columns if is_numeric_dtype(df[col]) and col not in ['Id', 'id']]

columnas_outliers = {}
total_filas = df.shape[0]

print("📊 Reporte estadístico de valores atípicos por columna (Método IQR):")

for col in variables_analisis:
    if col in df.columns:
        # Calcular Cuartiles (Q1 y Q3) y el Rango Intercuartílico (IQR)
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        # Definir los límites formales de Tukey
        limite_inferior = Q1 - 1.5 * IQR
        limite_superior = Q3 + 1.5 * IQR

        # Identificar las filas que se escapan de los límites
        filas_outliers = df[(df[col] < limite_inferior) | (df[col] > limite_superior)]
        cantidad = filas_outliers.shape[0]
        porcentaje = (cantidad / total_filas) * 100

        columnas_outliers[col] = cantidad
        print(f"  • {col:<20}: Detectados {cantidad:>2} outliers ({porcentaje:.1f}% del dataset).")

print('\n=== DETECCIÓN DE INCONSISTENCIAS NUMÉRICAS BASADAS EN LÍMITES LÓGICOS ===\n')

# Definir límites lógicos para el dataset Iris
# Estos límites se basan en las medidas reales de flores Iris registradas en la literatura botánica
limites_logicos = {
    'SepalLengthCm': {'min': 4.0, 'max': 8.5},      # Largo del sépalo: 4-8.5 cm típicamente
    'SepalWidthCm': {'min': 2.0, 'max': 4.5},       # Ancho del sépalo: 2-4.5 cm típicamente
    'PetalLengthCm': {'min': 1.0, 'max': 7.0},      # Largo del pétalo: 1-7 cm típicamente
    'PetalWidthCm': {'min': 0.1, 'max': 2.5}        # Ancho del pétalo: 0.1-2.5 cm típicamente
}

inconsistencias_num = {}

for col in variables_analisis:
    col_inconsistencias = []
    if col in limites_logicos:
        limite_min = limites_logicos[col].get('min')
        limite_max = limites_logicos[col].get('max')

        # Chequear valores bajo el mínimo
        if limite_min is not None:
            invalid_min = df[df[col] < limite_min]
            if not invalid_min.empty:
                col_inconsistencias.append(f"Valores menores que el mínimo lógico ({limite_min}): {invalid_min[col].unique()}")

        # Chequear valores sobre el máximo
        if limite_max is not None:
            invalid_max = df[df[col] > limite_max]
            if not invalid_max.empty:
                col_inconsistencias.append(f"Valores mayores que el máximo lógico ({limite_max}): {invalid_max[col].unique()}")

    if col_inconsistencias:
        inconsistencias_num[col] = col_inconsistencias

if not inconsistencias_num:
    print("✅ No se detectaron inconsistencias numéricas basadas en los límites lógicos definidos.")
else:
    print("⚠️ Se detectaron las siguientes inconsistencias numéricas:")
    for col, errores in inconsistencias_num.items():
        print(f"\n  • Columna '{col}':")
        for error in errores:
            print(f"    - {error}")
    print("\n💡 Estrategia propuesta para inconsistencias numéricas:")
    print("   Las inconsistencias detectadas deben ser:")
    print("   1. **Verificadas:** Confirmar si son datos legítimos o errores de medición.")
    print("   2. **Tratadas:** Dependiendo de la validación, se pueden:")
    print("      - Conservar si representan flores genuinamente atípicas.")
    print("      - Reemplazar por NaN si son claros errores de medición.")
    print("      - Eliminar solo en casos extremos donde el dato es inviable biológicamente.")


print('\n📈 Generando Boxplots explicativos para Outliers...')

# Cuadrícula de gráficos de caja para visualizar la dispersión
num_vars = len(variables_analisis)
num_cols = 2  # Número de columnas para los subplots
num_rows = (num_vars + num_cols - 1) // num_cols # Calcular el número de filas

fig, axes = plt.subplots(num_rows, num_cols, figsize=(num_cols * 5, num_rows * 4))
axes = axes.flatten() # Aplanar el array de axes para fácil iteración

for i, col in enumerate(variables_analisis):
    if col in df.columns:
        sns.boxplot(y=df[col], ax=axes[i], color='salmon', width=0.4)
        axes[i].set_title(f'Outliers en {col.replace("_", " ")}')
        axes[i].set_ylabel(f'{col} (cm)')

# Ocultar los ejes vacíos si hay más subplots que variables
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()

print("\n💡 Estrategia de Data Science adoptada para Outliers (Dataset Iris):")
print("   👉 Conservación Intencional: Los valores atípicos en el Iris pueden representar:")
print("      - Variabilidad natural dentro de cada especie")
print("      - Diferencias legítimas entre las 3 especies de Iris")
print("      - Especímenes únicos que son informativos para el modelo")
print("   👉 Mitigación: Se conservan todos los datos. Los algoritmos de clasificación")
print("      (KNN, Decision Tree, Logistic Regression) pueden manejar bien la variabilidad,")
print("      especialmente con StandardScaler para normalización, verificando antes el balance entre clases.")
print('-' * 80)
```

```python
print('=== Correlación entre Variables ===\n')

# Seleccionar solo columnas numéricas para el cálculo de correlación y excluir 'Id'
df_numeric = df.drop(columns=['Id', 'Species'], errors='ignore').select_dtypes(include=[np.number])

# Calcular la matriz de correlación
matriz_correlacion = df_numeric.corr()

# Visualizar la matriz de correlación con un mapa de calor
plt.figure(figsize=(10, 8))
sns.heatmap(matriz_correlacion, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5, 
            cbar_kws={'label': 'Correlación de Pearson'}, vmin=-1, vmax=1)
plt.title('Matriz de Correlación de Variables Numéricas (Iris)')
plt.tight_layout()
plt.show()

print('📊 Resumen de Correlaciones entre Variables del Iris:')
print()

# Analizar correlaciones más altas (excluyendo la diagonal)
correlaciones_altas = []
for i in range(len(matriz_correlacion.columns)):
    for j in range(i+1, len(matriz_correlacion.columns)):
        col1 = matriz_correlacion.columns[i]
        col2 = matriz_correlacion.columns[j]
        valor_corr = matriz_correlacion.iloc[i, j]
        correlaciones_altas.append((col1, col2, valor_corr))

# Ordenar por valor absoluto de correlación
correlaciones_altas.sort(key=lambda x: abs(x[2]), reverse=True)

print('Correlaciones más significativas (ordenadas por fuerza):\n')
for col1, col2, corr_val in correlaciones_altas:
    interpretacion = "🔴 Muy fuerte" if abs(corr_val) > 0.8 else "🟠 Fuerte" if abs(corr_val) > 0.6 else "🟡 Moderada"
    print(f"  {interpretacion}: {col1:<20} ↔ {col2:<20} = {corr_val:>6.3f}")

print('\n--------------------------------------------------------------------------------')
```

**2. Correlación Sépalo-Sépalo (FUERTE: ~0.82)**
- **Hallazgo:** SepalLengthCm y SepalWidthCm tienen fuerte correlación positiva
- **Interpretación:** Flores con sépalos más largos tienden a tener sépalos más anchos
- **Implicación:** Nuevamente hay cierta redundancia entre estas dos características

**3. Correlación Sépalo-Pétalo (DÉBIL a MODERADA: 0.27 - 0.87)**
- **Hallazgo clave:** PetalLengthCm y SepalLengthCm están fuertemente correlacionadas (~0.87)
- **Interpretación:** Flores con sépalos largos generalmente tienen pétalos largos - esto sugiere que ambas dimensiones crecen proporcionalmente
- **Implicación:** Las 3 especies de Iris difieren en su tamaño general (pequeñas, medianas, grandes)

### Impacto en Clasificación

✅ **Aspecto Positivo:**
- La correlación fuerte entre características sugiere que el tamaño general es un buen discriminador entre especies
- Esto facilitará la separación de clases en el espacio de características

⚠️ **Consideración:**
- La multicolinealidad (r > 0.8) puede afectar modelos como Regresión Logística
- Modelos basados en árboles (Decision Tree, Random Forest) son robustos ante multicolinealidad
- Para modelos sensibles, se podría considerar eliminar una de las características altamente correlacionadas

### Conclusión
El dataset Iris tiene estructura de **correlación coherente**: las medidas morfológicas están conectadas de forma biológicamente sensible. Las especies probablemente se diferencian principalmente por tamaño general, haciendo este dataset ideal para problemas de clasificación.


## 🎯 Diagnóstico y Selección del Problema: Dataset Iris

### 1. Diagnóstico del Dataset y Desafíos Clave
Tras el análisis exploratorio inicial (EDA), se diagnosticaron las siguientes características en este conjunto de datos morfológicos:
* **Dimensiones Reducidas y Balance Simétrico:** El dataset presenta una estructura pequeña, limpia y perfectamente balanceada en sus clases (50 muestras por tipo de flor), sin problemas severos de valores nulos o atípicos de origen instrumental.
* **Solapamiento Lineal Parcial:** Mientras que una de las especies (*Iris-setosa*) es perfectamente separable de forma lineal basándose en la longitud y ancho de los pétalos, las otras dos especies (*Iris-versicolor* e *Iris-virginica*) presentan zonas de solapamiento en sus fronteras de distribución, lo que desafía a los clasificadores lineales simples.

### 2. Definición de la Problemática Seleccionada
* **Tipo de Problema:** Clasificación Supervisada Multiclase.
* **Objetivo Analítico:** Clasificar la especie exacta de la planta (Setosa, Versicolor o Virginica) a partir de las mediciones morfométricas de sus sépalos y pétalos.

### 3. Justificación de la Elección y Relevancia
* **Relevancia:** Funciona como el modelo base de validación taxonómica en botánica. Permite auditar y comparar la eficiencia de algoritmos de separación de fronteras antes de escalarlos a problemas de biodiversidad masivos.
* **Desafío Técnico:** A pesar de su aparente simplicidad, el desafío técnico radica en lograr una separación perfecta en la zona de solapamiento de clases sin caer en el sobreajuste (*overfitting*), sirviendo como el entorno ideal para evaluar la selección de hiperparámetros y la reducción de dimensiones (como PCA) en la Fase II.
