  * Chequear incosistencias en valores categóricos

```python
print('=== Detección de Inconsistencias en Columnas Categóricas ===')
print()

# Identificar columnas categóricas en el DataFrame
cols_categoricas = df.select_dtypes(include=['category', 'object']).columns

if cols_categoricas.empty:
    print("• No se encontraron columnas de tipo 'category' u 'object' en el DataFrame. ✅")
    print("  Por lo tanto, no hay inconsistencias categóricas que chequear en este paso.")
else:
    print(f"• Se encontraron las siguientes columnas categóricas: {list(cols_categoricas)}")
    print("  Realizando verificación de inconsistencias...")

    columnas_con_inconsistencias = []

    for col in cols_categoricas:
        # Convertir a string para aplicar strip y luego convertir de nuevo a category.
        # Esto ayuda a manejar la FutureWarning asegurando que las categorías se reinfieran de las cadenas limpias.
        df[col] = df[col].astype(str).str.strip().astype('category')

        # La normalización temporal para la búsqueda de inconsistencias
        serie_limpia = df[col].astype(str).str.capitalize() # Operar sobre la columna ya limpia y categorizada
        conteo = serie_limpia.value_counts()
        # Filtrar valores que aparecen más de una vez (categorías reales)
        valores_validos = conteo[conteo > 1].index.tolist()

        print(f'Columna: {col}')
        # Encontrar valores inconsistentes (aquellos que aparecen solo una vez después de la capitalización)
        inconsistent_values = serie_limpia[~serie_limpia.isin(valores_validos)].unique().tolist()

        if len(inconsistent_values) > 0:
            print(f'• Valores únicos detectados (aparecen solo una vez): {inconsistent_values}')
            print("  Ejemplos de filas con estos valores:")
            # Asegurarse de indexar con la versión capitalizada para encontrar las filas correspondientes
            display(df[df[col].astype(str).str.capitalize().isin(inconsistent_values)][[col]])
            columnas_con_inconsistencias.append(col)
        else:
            print(f'• Sin inconsistencias detectadas (todos los valores aparecen más de una vez) ✅')
        print('-' * 40)

    if len(columnas_con_inconsistencias) == 0:
        print("\n✅ Resumen: No se encontraron nuevas inconsistencias o todas las detectadas fueron abordadas.")
    else:
        print(f"\n⚠️ Resumen: Se encontraron inconsistencias en las siguientes columnas: {', '.join(columnas_con_inconsistencias)}.")
print('-' * 80)
```

```python
print('=== Histograma de Distribución de Variables ===\n')

# Crear una copia del DataFrame para trabajar y aplicar muestreo si es necesario
df_plot = df.copy()

# Si el dataset es muy grande, tomar una muestra para la visualización
# Esto ayuda a reducir el tiempo de renderizado y el uso de memoria para los gráficos.
if len(df_plot) > 100000:
    print(f"Dataset grande ({len(df_plot)} filas). Tomando una muestra de 10,000 filas para la visualización de histogramas.")
    df_plot = df_plot.sample(n=10000, random_state=42)

# Seleccionar solo columnas numéricas para los histogramas donde el cálculo de media y mediana es significativo.
# Excluir 'Customer ID' (ya que es un identificador) y cualquier columna 'Id'.
# excluyendo correctamente 'object', 'category' y 'datetime' types que no son adecuados para este tipo de análisis estadístico.
variables_a_graficar = [col for col in df_plot.select_dtypes(include=np.number).columns if col not in ['Customer ID', 'Id']]

num_vars = len(variables_a_graficar)
num_cols = 3  # Número de columnas para los subplots
num_rows = (num_vars + num_cols - 1) // num_cols if num_vars > 0 else 0 # Calcular el número de filas, manejar caso de 0 variables

# Ajustar la creación de subplots para casos con 0 o 1 variable
if num_vars == 0:
    print("No hay variables numéricas adecuadas para graficar histogramas con líneas de media/mediana.")
elif num_vars == 1:
    fig, axes = plt.subplots(1, 1, figsize=(num_cols * 5, 4)) # Un solo gráfico
    axes = [axes] # Convertir a lista para que sea iterable como el array aplanado
else:
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(num_cols * 5, num_rows * 4))
    axes = axes.flatten() # Aplanar el array de axes para fácil iteración

# Solo proceder con la creación de gráficos si hay variables que graficar
if num_vars > 0:
    for i, col in enumerate(variables_a_graficar):
        sns.histplot(df_plot[col], kde=True, ax=axes[i], color='skyblue', edgecolor='black', line_kws={'linewidth': 4, 'color': 'darkblue'}) # Increased KDE linewidth further
        axes[i].set_title(f'Distribución de {col.replace("_", " ").title()}')
        axes[i].set_xlabel(col.replace("_", " ").title())
        axes[i].set_ylabel('Frecuencia')
        # Añadir una línea vertical para la media y la mediana
        axes[i].axvline(df_plot[col].mean(), color='red', linestyle='dashed', linewidth=3, label='Media') # Increased linewidth further
        axes[i].axvline(df_plot[col].median(), color='green', linestyle='dotted', linewidth=3, label='Mediana') # Increased linewidth further
        axes[i].legend(loc='upper right') # Explicitly set legend location to avoid performance warning

    # Ocultar los ejes vacíos si hay más subplots que variables
    for j in range(num_vars, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()

print('\n💡 Interpretación de los Histogramas:')
print("   • **Quantity:** El histograma para 'Quantity' muestra una distribución fuertemente sesgada a la derecha, con la mayoría de las transacciones involucrando cantidades pequeñas. La presencia de valores negativos (indicando devoluciones o cancelaciones) también es visible. La media y la mediana difieren, confirmando el sesgo y la no normalidad (como también indicó el test de Shapiro-Wilk).")
print("   • **Price:** Similar a 'Quantity', la distribución de 'Price' también está fuertemente sesgada a la derecha, con la mayoría de los artículos teniendo precios bajos. Hay una cola larga hacia precios más altos y la presencia de valores negativos (probablemente errores o ajustes) en el eje x. La diferencia entre la media y la mediana y el resultado del test de Shapiro-Wilk confirman que no sigue una distribución normal.")
print('   • En ambos casos (Quantity y Price), la clara desviación de la normalidad sugiere que los modelos que asumen una distribución normal pueden no ser adecuados sin transformaciones de datos previas.')
print('--------------------------------------------------------------------------------')
```

```python
print('=== Verificación de Normalidad de Variables (Test de Shapiro-Wilk) ===\n')

from scipy.stats import shapiro

# Excluir 'Id', 'Customer ID' y 'quality' de las columnas a testear si están presentes
# y asegurar que solo se seleccionen columnas numéricas para el test de normalidad.
# 'Quantity' y 'Price' son las variables numéricas clave para análisis.
variables_para_normalidad = [col for col in df.select_dtypes(include=np.number).columns if col not in ['Customer ID', 'Id', 'quality']]

alpha = 0.05  # Nivel de significancia

print(f"Hipótesis nula (H0): La distribución de la variable es normal.\nNivel de significancia (alpha): {alpha}\n")

if not variables_para_normalidad:
    print("No se encontraron variables numéricas adecuadas para realizar el test de Shapiro-Wilk.")
else:
    for col in variables_para_normalidad:
        # Eliminar NaN antes de aplicar Shapiro-Wilk, ya que no maneja nulos
        data = df[col].dropna()

        # El test de Shapiro-Wilk requiere al menos 3 puntos de datos. Si hay pocos, se salta.
        if len(data) < 3:
            print(f'\n--- Columna: {col} ---')
            print(f'No hay suficientes datos (menos de 3) para realizar el test de Shapiro-Wilk. Saltando.')
            continue

        # Para datasets grandes, el test de Shapiro-Wilk es computacionalmente intensivo y puede ser demasiado sensible.
        # Se tomará una muestra para el test si la columna tiene más de 5000 observaciones.
        sample_limit = 5000
        if len(data) > sample_limit:
            print(f'\n--- Columna: {col} ---')
            print(f'La columna tiene {len(data)} observaciones. Se ha tomado una muestra de {sample_limit} para el test de Shapiro-Wilk.')
            data = data.sample(n=sample_limit, random_state=42)
        elif len(data) >= 3: # If data is smaller than sample_limit but still large enough for the test
            print(f'\n--- Columna: {col} ---')
            print(f'La columna tiene {len(data)} observaciones. Se utilizarán todos los datos para el test de Shapiro-Wilk.')

        stat, p = shapiro(data)
        print(f'Estadístico de Shapiro-Wilk: {stat:.4f}')
        print(f'Valor p: {p:.4e}') # Formato exponencial para p-values pequeños
        if p > alpha:
            print('Conclusión: No se puede rechazar la hipótesis nula (H0). La columna parece tener una distribución normal. ✅')
        else:
            print('Conclusión: Se rechaza la hipótesis nula (H0). La columna NO sigue una distribución normal. ❌')

print('\n--------------------------------------------------------------------------------')
print('💡 Nota: Para datasets grandes (ej. > 5000 muestras), el test de Shapiro-Wilk puede ser muy sensible a pequeñas desviaciones de la normalidad. En esos casos, la inspección visual (histogramas, Q-Q plots) o otros tests como Kolmogorov-Smirnov pueden ser más útiles.')
```

## Outliers



```python
print('=== Análisis y detección de valores atípicos y consistencia numérica ===\n')

# Excluimos columnas identificadoras y de fecha del análisis de outliers numéricos
# 'Customer ID' es un identificador numérico.
# 'InvoiceDate' es una fecha y no debe analizarse para outliers numéricos.
variables_analisis = [col for col in df.select_dtypes(include=np.number).columns if col not in ['Customer ID']]

columnas_outliers = {}
total_filas = df.shape[0]

print("📊 Reporte estadístico de valores atípicos por columna (Método IQR):")

for col in variables_analisis:
    if col in df.columns:
        # Calcular Cuartiles (Q1 y Q3) y el Rango Intercuartílico (IQR)
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        # Definir los límites formales de Tukey
        limite_inferior = Q1 - 1.5 * IQR
        limite_superior = Q3 + 1.5 * IQR

        # Identificar las filas que se escapan de los límites
        filas_outliers = df[(df[col] < limite_inferior) | (df[col] > limite_superior)]
        cantidad = filas_outliers.shape[0]
        porcentaje = (cantidad / total_filas) * 100

        columnas_outliers[col] = cantidad
        print(f"  • {col:<20}: Detectados {cantidad:>2} outliers ({porcentaje:.1f}% del dataset). ")

# ==============================================================================
# DETECCIÓN DE INCONSISTENCIAS NUMÉRICAS BASADAS EN LA NATURALEZA
# ==============================================================================

print('\n=== DETECCIÓN DE INCONSISTENCIAS NUMÉRICAS BASADAS EN LÍMITES LÓGICOS ===\n')

# Definir límites lógicos para las columnas numéricas relevantes del dataset de Retail
limites_logicos = {
    'Quantity': {'min': -100000.0, 'max': 100000.0}, # Cantidad puede ser negativa (devoluciones), y un máximo arbitrario
    'Price': {'min': 0.0, 'max': 100000.0} # Precio no puede ser negativo, y un máximo arbitrario
}

inconsistencias_num = {}

for col in variables_analisis: # Iteramos solo sobre las variables de análisis (Quantity, Price)
    col_inconsistencias = []
    if col in limites_logicos: # Asegurarse de que la columna tiene límites definidos
        limite_min = limites_logicos[col].get('min')
        limite_max = limites_logicos[col].get('max')

        # Chequear valores bajo el mínimo
        if limite_min is not None:
            invalid_min = df[df[col] < limite_min]
            if not invalid_min.empty:
                col_inconsistencias.append(f"Valores menores que el mínimo lógico ({limite_min}): {invalid_min[col].unique()}")

        # Chequear valores sobre el máximo
        if limite_max is not None:
            invalid_max = df[df[col] > limite_max]
            if not invalid_max.empty:
                col_inconsistencias.append(f"Valores mayores que el máximo lógico ({limite_max}): {invalid_max[col].unique()}")

    if col_inconsistencias:
        inconsistencias_num[col] = col_inconsistencias

if not inconsistencias_num:
    print("✅ No se detectaron inconsistencias numéricas basadas en los límites lógicos definidos.")
else:
    print("⚠️ Se detectaron las siguientes inconsistencias numéricas:")
    for col, errores in inconsistencias_num.items():
        print(f"\n  • Columna '{col}':")
        for error in errores:
            print(f"    - {error}")
    print("\n💡 Estrategia propuesta para inconsistencias numéricas:")
    print("   Las inconsistencias detectadas (ej. valores fuera de rangos plausibles) deben ser:")
    print("   1. **Verificadas:** Confirmar si son errores de entrada de datos o representaciones válidas de casos extremos.")
    print("   2. **Tratadas:** Dependiendo de la validación, se pueden: ")
    print("      - Reemplazar por NaN (para imputación posterior si se considera un error). ")
    print("      - Corregir a un valor lógico (ej. ajustar a un límite si es un error de medición). ")
    print("      - Eliminar la fila si el valor es crítico y la corrección inviable, aunque esto se hace con precaución.")


print('\n📈 Generando Boxplots explicativos para Outliers...')

# Cuadrícula de gráficos de caja para visualizar la dispersión
num_vars = len(variables_analisis)
num_cols = 2  # Número de columnas para los subplots (para Quantity y Price)
num_rows = (num_vars + num_cols - 1) // num_cols # Calcular el número de filas

fig, axes = plt.subplots(num_rows, num_cols, figsize=(num_cols * 4, num_rows * 4))
axes = axes.flatten() # Aplanar el array de axes para fácil iteración

for i, col in enumerate(variables_analisis):
    if col in df.columns:
        sns.boxplot(y=df[col], ax=axes[i], color='salmon', width=0.4)
        axes[i].set_title(f'Outliers en {col.replace("_", " ").title()}') # Formatear el título
        axes[i].set_ylabel('')

# Ocultar los ejes vacíos si hay más subplots que variables
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()

print("\n💡 Estrategia de Data Science adoptada para Outliers:")
print("   👉 Conservación Intencional: Para el análisis de transacciones, los valores atípicos (grandes cantidades, precios inusuales, o cantidades negativas por devoluciones) son fundamentales y no deben eliminarse sin un análisis profundo, ya que pueden representar eventos significativos (ej. compras mayoristas, errores de stock, clientes problemáticos). ")
print("   👉 Mitigación: Se utilizarán algoritmos robustos a outliers o técnicas de transformación (ej. winsorización o logarítmicas si la distribución lo permite) en etapas posteriores del modelado, o se manejarán explícitamente como características en el análisis RFM.")
print('-' * 80)
```

### Implicaciones para Modelado Predictivo

✅ **Aspectos Positivos:**
- La ausencia de una fuerte correlación lineal entre `Quantity` y `Price` sugiere que ambas variables aportan información independiente al modelo, lo cual es generalmente deseable. No hay una redundancia de información lineal entre estas dos características.

⚠️ **Consideraciones:**
- Aunque la correlación lineal es mínima, esto no descarta la posibilidad de relaciones no lineales más complejas entre `Quantity` y `Price` que podrían ser relevantes para el modelado. Un modelo de regresión lineal simple que intente predecir uno a partir del otro con solo estas dos variables sería ineficaz.
- Para modelos que dependen de la independencia de las variables predictoras (como la regresión lineal), la baja correlación entre `Quantity` y `Price` es beneficiosa, ya que evita problemas de multicolinealidad entre ellas.

### Conclusión
El dataset de Transacciones de Retail Online indica una **correlación lineal casi inexistente** entre la cantidad de productos comprados y su precio unitario. Esta independencia lineal entre `Quantity` y `Price` es un hallazgo importante que debe ser considerado en cualquier esfuerzo de modelado posterior, sugiriendo que ambas variables pueden ser tratadas como características distintas y valiosas sin preocuparse por la multicolinealidad lineal entre ellas.
